```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
data = {
    'Days': ['Mon', 'Tue', 'Wed', 'Thur', 'Fri', 'Saty', 'Sun'],
    'Study_Hours': [8, 7, 6, 5, 4, 3, 2],
    'Marks': [90, 85, 80, 75, 70, 65, 60],
    'Attendance': [95, 90, 85, 80, 75, 70, 65],
    'Sleep_Hours': [7, 6, 6, 7, 8, 6, 5],
    'Assignment_Completed': [1,2,1,3,3,2,1]
}
df = pd.DataFrame(data)
df
```

# LINE CHART (TREND)

```python
plt.plot(df['Days'], df['Study_Hours'], label='Study Hours', marker='*')
plt.title('Ansh')
plt.xlabel('Days')
plt.ylabel('Study_Hours')
plt.legend()
plt.grid(True)
plt.show()
```

**Advanced Line Chart**

```python
plt.figure(figsize=(7, 6))
plt.plot(df['Days'], df['Study_Hours'], label='Study Hours', marker='o', linestyle='-',color='blue',)
plt.plot(df['Days'], df['Marks'], label='Marks', marker='s', linestyle='--',color='red')
plt.legend()
plt.grid(True)
plt.title('Ansh')
plt.show()
```

# BAR CHART

```python
plt.bar(df['Days'], df['Marks'], label='Marks',color='green')
plt.title('Ansh')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.legend()
plt.show()
```

**Advanced Bar Graph**

```python
bars = plt.bar(df['Days'], df['Marks'], label='Marks',color='red')
for bar in bars:
    y = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, y, str(y),  ha='center', va='bottom')
plt.title('Ansh')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.grid(axis = 'y')
plt.legend()
```

# Histogram Distribution

```python
data = {
    'marks': [90, 85, 80, 75, 70, 65, 60, 67, 82, 79, 99, 94, 69, 93, 88, 84]
}
df = pd.DataFrame(data)
plt.hist(df['marks'], bins=5, edgecolor='black',alpha= 0.70)
plt.title('Ansh')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.show()
```

**Advanced Histogram**

```python
import matplotlib.pyplot as plt
import numpy as np
marks = [90, 85, 80, 75, 70, 65, 60, 67, 82, 79, 99, 94, 69, 93, 88, 84]
#Calculating mean
mean_value = np.mean(marks)

plt.hist(marks, bins=5, edgecolor='black', alpha=0.7)

#mean line
plt.axvline(mean_value, color='red', linestyle='dashed', linewidth=2,label='Mean')

plt.title('Ansh')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

# Scatter Plot

```python
import matplotlib.pyplot as plt
plt.scatter(df['Days'], df['Marks'], label='Marks',color='green')
plt.title('Ansh')
plt.xlabel('Days')
plt.ylabel('Marks')
plt.legend()
plt.show()
```

**Advanced Saccter Plot**

```python
df['Result'] = ['Pass' if marks >= 70 else 'Fail' for marks in df['Marks']]
colors = ['red' if result == 'Fail' else 'green' for result in df['Result']]
plt.scatter(df['Study_Hours'], df['Marks'], label='Marks',c=colors)
plt.title('Ansh')
plt.xlabel('Study_Hours')
plt.ylabel('Marks')
plt.legend()
plt.show()
```
